```python
import matplotlib.pyplot as plt
```

```python
sns.histplot(df['petal_length'])
plt.title("histogram - petal_length")
plt.show()
```

```python
df.shape
```

```python
sns.histplot(data = df, x = 'petal_length',bins = 5,color = 'blue',kde=True)
plt.title("histogram - petal_length")
plt.show()
```

```python
#barplot - 2 cols - num- cat

sns.barplot(data = df, x ='species',y ='petal_length',hue ='species',palette='dark')
plt.show()
```

```python
sns.boxplot(data = df, x ='species',y ='petal_length',hue ='species') #only numerical
plt.show()
```

```python
#scatter plot - num vs num
sns.scatterplot(data = df, x = 'petal_length',y = 'sepal_length',hue ='species')
plt.show()
```
