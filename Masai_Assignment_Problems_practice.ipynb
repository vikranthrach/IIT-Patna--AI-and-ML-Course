**Problem Statement:**

A data analyst needs to analyze customer order data from an e-commerce database. You are given a database with two tables: customers (columns: customer_id INTEGER PRIMARY KEY AUTOINCREMENT, name TEXT, city TEXT) and orders (columns: order_id INTEGER PRIMARY KEY AUTOINCREMENT, customer_id INTEGER FOREIGN KEY, amount REAL, order_date DATE). Write a complete Python script that:

Creates both tables (if they do not exist)

Inserts sample customer and order data

Executes a query to retrieve customers from New York who placed orders over 100 dollars, displaying customer name, city, and order amount

Saves the query results to a Pandas DataFrame

Outputs the results

The script should demonstrate proper database connection handling, table creation with constraints, batch inserts, and query execution.

Constraints & Requirements:

Use SQLite3 with a file-based database named ecommerce.db
Use executemany() for batch inserts
Include NOT NULL and UNIQUE constraints where appropriate
Write a single SQL query that combines filtering conditions (WHERE, AND)
Close the database connection after operations complete
Include comments explaining non-obvious steps
Example Input / Expected Output:

Input: **bold text** Sample data with 3 customers and 4 orders; at least one customer from New York with an order over $100
Expected Output: A DataFrame (and printed output) showing the customer name, city, and order amount for qualifying records

```python
# Import required libraries

import sqlite3

import pandas as pd



# ---------------------------------------------------

# STEP 1: Create/connect to database

# ---------------------------------------------------



# This creates a database file named ecommeerce.db

connection = sqlite3.connect("ecommerce.db")



# Cursor is used to execute SQL commands

cursor = connection.cursor()



print("Database connected successfully!")



# ---------------------------------------------------

# STEP 2: Create tables with constraints

# ---------------------------------------------------



# Create Customers table

cursor.execute("""

CREATE TABLE IF NOT EXISTS customers (

    customer_id INTEGER PRIMARY KEY,

    name TEXT NOT NULL,

    city TEXT NOT NULL

)

""")



# Create Orders table

cursor.execute("""

CREATE TABLE IF NOT EXISTS orders (

    order_id INTEGER PRIMARY KEY,

    customer_id INTEGER,

    product TEXT NOT NULL,

    amount REAL CHECK(amount > 0),



    FOREIGN KEY(customer_id)

    REFERENCES customers(customer_id)

)

""")



print("Tables created successfully!")



# ---------------------------------------------------

# STEP 3: Insert sample data using executemany()

# ---------------------------------------------------



# Sample customer data

customers_data = [

    (1, "Alice", "New York"),

    (2, "Bob", "Chicago"),

    (3, "Charlie", "New York"),

    (4, "David", "Boston")

]



# Insert customers

cursor.executemany("""

INSERT OR REPLACE INTO customers

(customer_id, name, city)

VALUES  (?, ?, ?)""", customers_data)



# Sample orders data

orders_data = [

    (101, 1, "Laptop", 1200),

    (102, 2, "Mouse", 25),

    (103, 3, "Keyboard", 150),

    (104, 1, "Monitor", 300),

    (105, 4, "USB Cable", 15)

]



# Insert orders

cursor.executemany("""

INSERT OR REPLACE INTO orders

(order_id, customer_id, product, amount)

VALUES (?, ?, ?, ?)

""", orders_data)



# Save changes

connection.commit()



print("Sample data inserted successfully!")



# ---------------------------------------------------

# STEP 4: Query customers from New York

#         with orders over 100

# ---------------------------------------------------



query = """

SELECT

    customers.customer_id,

    customers.name,

    customers.city,

    orders.product,

    orders.amount

FROM customers

JOIN orders

ON customers.customer_id = orders.customer_id

WHERE customers.city = 'New York'

AND orders.amount > 100

"""



# ---------------------------------------------------

# STEP 5: Load results into Pandas DataFrame

# ---------------------------------------------------



df = pd.read_sql_query(query, connection)



print("\nQuery Results:")

print(df)



# ---------------------------------------------------

# STEP 6: Close the connection

# ---------------------------------------------------



cursor.close()

connection.close()



print("\nDatabase connection closed!")
```

**Pandas Problem:**
You are working with the Titanic dataset, which has 891 rows and columns including sex, pclass, age, fare, embarked, and survived. The dataset has 177 missing values in the age column (approximately 20% of records).

Your task is to write a complete Python script using Pandas that performs the following data analysis pipeline:

Requirements:

Load the Titanic dataset from a CSV file called titanic.csv using read_csv
Filter passengers who are female AND in pclass 1 or pclass 2 (use isin for the pclass check)
From this filtered group, fill missing age values with the value 30 using fillna
Sort the filtered DataFrame by fare in descending order, then reset the index so the index values are sequential from 0
Add a new column called fare_group that assigns the string 'high' to all rows (you may assign a constant string value to the new column)
Save the final result to a new CSV file called female_upper_class.csv

```python

#Import Pandas

import pandas as pd

#---------------------------------------------
# STEP 1: Load Titanic dataset from CSV file
#---------------------------------------------

titanic=pd.read_csv("titanic.csv")

titanic
#---------------------------------------------------
# STEP 2:Filter Passengers
# Female passengers
# pclass is either 1 or 2
#---------------------------------------------------
passengers=titanic[(titanic["Sex"]=="female") & (titanic["Pclass"].isin([1,2]))]
passengers
#-----------------------------------
# STEP 3: Filling missing age values with 30
#--------------------------------------------

passengers=passengers.fillna(30)
passengers

#-------------------------------------------------------
# STEP 4: Sort by Fare in descending order
# Then reset index
#---------------------------------------------------------
passengers=passengers.sort_values(by="Fare",ascending=False).reset_index(drop=True)
passengers
#-------------------------------------------------------
# STEP 5: Add new column fare_group
#---------------------------------------------------------

passengers["fare_group"]="high"

#-------------------------------------------------------
# STEP 6: Save final result to new CSV file
#---------------------------------------------------------

passengers.to_csv("female_upper_class.csv",index=False)
passengers
```

Load the CSV into a Pandas DataFrame.

Rename the column score to final_score and grade to letter_grade without modifying the original DataFrame (save the result to a new variable).

Using iloc, access the first 3 rows and all columns. Then using loc, access rows with index labels 1 and 3, and only the name and grade columns.

Walkthrough:

pd.read_csv('students.csv') loads the file and auto-assigns a default integer index starting at 0.
df.shape returns (4, 4) — 4 rows and 4 columns.
df.dtypes shows that name and grade are object (string) type, age is int64, and score is float64.
df.describe() only shows age and score because it only operates on numerical data. The two object columns (name, grade) are excluded.
df.rename(columns={...}) without inplace=True returns a new DataFrame. The original df is unchanged.
df_renamed[['name', 'final_score', 'letter_grade']] uses a list inside double brackets to select multiple columns, returning a DataFrame.
df.iloc[0:3] accesses rows at integer positions 0, 1, 2. Position 3 is exclusive.
df.loc[[1, 3], ['name', 'grade']] accesses rows by their actual index labels 1 and 3. Since the default index is 0–3, these correspond to Bob and Diana respectively.

```python
import pandas as pd
 #(Answer provided in Assignment)

# Step 1: Load the CSV
df = pd.read_csv('students.csv')

# Step 2: Inspect structure
print(df.shape)        # (4, 4)
print(df.columns)      # Index(['name', 'age', 'score', 'grade'], dtype='object')
print(df.dtypes)
# name     object
# age       int64
# score   float64
# grade    object

# Step 3: Describe
print(df.describe())
# Only age and score appear — these are numerical (int64 and float64).
# 'name' and 'grade' are object (string) columns and are excluded from df.describe().

# Step 4: Rename without modifying original
df_renamed = df.rename(columns={'score': 'final_score', 'grade': 'letter_grade'})
# df still has original column names; df_renamed has the new names

# Step 5: Select specific columns from renamed DataFrame
result = df_renamed[['name', 'final_score', 'letter_grade']]
print(result)

# Step 6a: iloc — first 3 rows, all columns
print(df.iloc[0:3])   # Rows at positions 0, 1, 2 — position 3 is excluded (iloc is exclusive at end)

# Step 6b: loc — rows with labels 1 and 3 and specific columns
print(df.loc[[1, 3], ['name', 'grade']])
# loc uses actual index labels (0-3 after read_csv) and is inclusive

```
